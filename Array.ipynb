# Array

Array  is an special variable it  is store more than one value or items at a time

```python
cars=['BMW',"Mersdese","Bugati","Odee"]
print(cars)
```

'len' function is used to find the length of array



```python
print(type(cars))
```

```python
x=cars[0]
x
```

```python
x=cars[1:4]
x
```

```python
x=len(cars)
x
```

```python
x=cars[3]
x
```

```python
print(type(cars))
```

 print each items in array

Add  Element Method 

```python
# You can use the 'append' method to add an element in a array
# append method use to add only one item at a time
```

```python
cars.append('Honda')

print(cars)
```

```python
print(cars)
```

Removing element in the array

<!-- We can use 'pop' method to removing the element  -->
<!-- but the pop method remove only one item at the end -->

```python
# to removing element at the end than we can use pop method
```

```python
cars.pop()
```

```python
print(cars)
```

Also we can use an another method to remove element from an array
'remove'  method can remove the element you want

```python
cars.remove('Bugati')
```

```python
print(cars)
```
